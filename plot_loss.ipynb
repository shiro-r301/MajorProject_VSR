```python
import re

losses = []
mses = []
percs = []
frames = []

pattern = r"loss ([\d.]+) \(mse ([\d.]+), perc ([\d.]+), frame ([\d.]+)\)"

with open("train_debug.log", "r") as f:
    for line in f:
        match = re.search(pattern, line)

        if match:
            loss, mse, perc, frame = map(float, match.groups())

            losses.append(loss)
            mses.append(mse)
            percs.append(perc)
            frames.append(frame)

print(len(mses))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(40,10))
plt.plot(losses, label="Loss")
plt.plot(mses, label="MSE")
plt.plot(percs, label="Perceptual")
plt.plot(frames, label="Frame")

plt.xlabel("Training Step")
plt.ylabel("Value")
plt.title("Training Losses")
plt.legend()
plt.grid(True)
plt.show()
```
