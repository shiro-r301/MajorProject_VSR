```python
import re

with open("train_debug (2).log", "r") as f:
    text = f.read()

losses = re.findall(
    r"\[Loss\].*?\bmean=([0-9]+(?:\.[0-9]+)?)",
    text
)

losses = [float(x) for x in losses]

print(len(losses))
```

```python
import matplotlib.pyplot as plt

idx = range(len(losses))

plt.plot(idx, losses)
plt.figure(figsize=(10,20))
plt.xlabel("Step")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.grid(True)
plt.show()
```
